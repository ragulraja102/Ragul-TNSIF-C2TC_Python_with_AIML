# Unsupervised Learning — Core Concepts Made Simple

**Unsupervised Learning** is like giving a computer a set of unsorted items without any labels or titles and asking it to discover hidden patterns, group similar items together, or simplify complex data on its own.

---

### Core Concepts Reference:

| Technique / Concept | Simple Analogy | Goal |
| :--- | :--- | :--- |
| **K-Means Clustering** | Sorting mixed items into $K$ distinct piles based on similarity | Group data into $K$ specified clusters automatically |
| **Hierarchical Clustering** | Creating a family tree (dendrogram) showing how smaller groups merge | Build a step-by-step tree structure of connections |
| **PCA (Principal Component Analysis)** | Taking a 2D shadow photo of a 3D object to capture its main shape | Compress many columns into fewer columns while keeping key patterns |
| **t-SNE** | Creating a 2D map of a complex high-dimensional landscape | Reduce complex data into 2D specifically for easy visual plotting |
| **Apriori Algorithm** | Spotting "frequently bought together" items at a local market | Discover hidden association rules (If item X is bought, then item Y is bought) |

---

#### 1. Clustering Techniques (K-Means & Hierarchical)

```python
import pandas as pd
from sklearn.cluster import KMeans, AgglomerativeClustering

# Step 1: Create a simple dataset of student Marks and Attendance %
data = {
    "Marks": [85, 90, 20, 15, 88, 25],
    "Attendance": [90, 95, 30, 25, 92, 35]
}
df = pd.DataFrame(data)

# --- A. K-MEANS CLUSTERING ---
# Group students into K=2 clusters (e.g., High Performers vs Low Performers)
kmeans = KMeans(n_clusters=2, random_state=42)

# fit_predict learns the groups and assigns a cluster label (0 or 1) to each student
df["KMeans_Group"] = kmeans.fit_predict(df[["Marks", "Attendance"]])

print("--- K-Means Grouping ---")
print(df[["Marks", "Attendance", "KMeans_Group"]])


# --- B. HIERARCHICAL CLUSTERING ---
# Group students by building a step-by-step tree structure until 2 groups remain
hierarchical = AgglomerativeClustering(n_clusters=2)

# Assign cluster labels based on the hierarchy tree
df["Hierarchical_Group"] = hierarchical.fit_predict(df[["Marks", "Attendance"]])

print("\n--- Hierarchical Grouping ---")
print(df[["Marks", "Attendance", "Hierarchical_Group"]])
```

### 2. Dimensionality Reduction (PCA & t-SNE)

```python
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

# Step 1: Create a dataset with 4 subject marks (4 dimensions/columns)
data = {
    "Math": [90, 85, 20, 30],
    "Science": [92, 88, 25, 35],
    "English": [78, 80, 40, 45],
    "History": [82, 85, 50, 52]
}
df = pd.DataFrame(data)

# --- A. PCA (Compress 4 subject columns down to 2 summary columns) ---
pca = PCA(n_components=2)
pca_result = pca.fit_transform(df)

# Convert the compressed array back into a DataFrame
df_pca = pd.DataFrame(pca_result, columns=["PCA_Column_1", "PCA_Column_2"])
print("--- PCA (Reduced from 4 columns to 2) ---")
print(df_pca)


# --- B. t-SNE (Compress 4 columns down to 2 specifically for plotting/mapping) ---
# perplexity=2 sets how many neighbors each point looks at
tsne = TSNE(n_components=2, perplexity=2, random_state=42)
tsne_result = tsne.fit_transform(df)

df_tsne = pd.DataFrame(tsne_result, columns=["tSNE_Column_1", "tSNE_Column_2"])
print("\n--- t-SNE (2D Map Coordinates) ---")
print(df_tsne)
```

### 3. Association Rule Mining (Apriori Algorithm)

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
import warnings
warnings.filterwarnings("ignore")

# Step 1: List of items bought together in 5 different transactions
transactions = [
    ["Notebook", "Pen", "Eraser"],
    ["Notebook", "Pen"],
    ["Notebook", "Scale"],
    ["Pen", "Eraser"],
    ["Notebook", "Pen", "Scale"]
]

# Step 2: Convert text list into a True/False table (One-Hot Encoding)
te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
df_basket = pd.DataFrame(te_ary, columns=te.columns_)

# Step 3: Find items appearing together in at least 40% of transactions (min_support=0.4)
frequent_items = apriori(df_basket, min_support=0.4, use_colnames=True)
print("--- Frequently Bought Itemsets ---")
print(frequent_items)

# Step 4: Extract rules (e.g., If buying a Notebook, how likely is buying a Pen?)
rules = association_rules(frequent_items, metric="confidence", min_threshold=0.6)
print("\n--- Association Rules (IF X -> THEN Y) ---")
print(rules[["antecedents", "consequents", "support", "confidence"]])


# The warnings printed at the bottom are just standard Python/Jupyter deprecation warnings about datetime.utcnow() due to Outdated Library Function or Internal Colab Logger
```
